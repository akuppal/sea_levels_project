```python
import pandas as pd
import altair as alt
import plotly.graph_objects as go
import numpy as np
```

```python
industry_df = pd.read_csv('Industry_Carbon_Emission.csv')
sea_df = pd.read_csv('Sea_Levels.csv')
```

**Proposition: Increase in use of water transport in global trade increases net carbon emissions, resulting in higher sea levels.**

```python
# Columns to keep
year_columns = [str(year) for year in range(1995, 2019)]
useful_columns = ['Country', 'Industry', 'Unit'] + year_columns

# Filter and subset
water_transport_df = industry_df[
    (industry_df['Industry'] == 'Water transport')
][useful_columns]
water_transport_df
```

```python
# Step 1: Rename columns to avoid confusion (optional but helpful)
sea_level_df = sea_df.rename(columns={'Measure': 'Location'})

# Step 2: Keep only relevant columns
sea_level_df = sea_level_df[['Date', 'Location', 'Value', 'Unit']].copy()

# Step 3: Convert Date to datetime
sea_level_df['Date'] = pd.to_datetime(sea_level_df['Date'].str.replace('D', ''))
sea_level_df.sort_values(['Location'])
```

```python
water_transport_df['Unit'].unique()
```

```python
water_transport_df = water_transport_df.groupby('Unit').sum().iloc[:,2:]
```

```python
# Plot emission (ppm) due to water transport per year
emission_df = water_transport_df.iloc[1:]

emission_df = emission_df.melt(
    var_name='Year',
    value_name='emission'
)
emission_df['Year'] = emission_df['Year'].astype(int)
alt.Chart(emission_df).mark_line(point=True).encode(
    x=alt.X('Year:O', title='Year', axis=alt.Axis(labelAngle=0)),
    y=alt.Y('emission:Q',
            title='CO₂ Emission (PPM)',
            scale=alt.Scale(domain=[490, 900])),
    tooltip=['Year', 'emission']
).properties(
    title='Water Transport CO₂ Emissions (1995–2018)',
    width=600,
    height=400
).interactive()
```

```python
sea_level_df['Year'] = pd.to_datetime(sea_level_df['Date']).dt.year
yearly_sea_levels = sea_level_df.groupby('Year', as_index=False)['Value'].mean()
yearly_sea_levels = yearly_sea_levels[(yearly_sea_levels['Year']>=1995) & (yearly_sea_levels['Year']<=2018)]
yearly_sea_levels.columns = ['Year', 'AvgSeaLevel']
```

```python
alt.Chart(yearly_sea_levels).mark_line(point=True).encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('AvgSeaLevel:Q', title='Average Sea Level (mm)'),
    tooltip=['Year', 'AvgSeaLevel']
).properties(
    title='Average Global Sea Level Change by Year',
    width=600,
    height=400
).interactive()
```

```python
# Make sure data is sorted
emission_df = emission_df.sort_values('Year')
yearly_sea_levels = yearly_sea_levels.sort_values('Year')

# Apply 3-year centered rolling average
emission_df['emission_smoothed'] = emission_df['emission'].rolling(window=3, center=True).mean()
yearly_sea_levels['AvgSeaLevel_smoothed'] = yearly_sea_levels['AvgSeaLevel'].rolling(window=3, center=True).mean()
```

```python
# --- CO2 Emissions Trace (left Y-axis) ---
emission_trace = go.Scatter(
    x=emission_df['Year'],
    y=emission_df['emission'],
    name='CO₂ Emissions (PPM)',
    yaxis='y1',
    mode='lines+markers',
    line=dict(color='green')
)

# --- Sea Level Trace (right Y-axis) ---
sea_level_trace = go.Scatter(
    x=yearly_sea_levels['Year'],
    y=yearly_sea_levels['AvgSeaLevel'],
    name='Sea Level (mm)',
    yaxis='y2',
    mode='lines+markers',
    line=dict(color='blue')
)

# --- Combine the Traces ---
fig = go.Figure(data=[emission_trace, sea_level_trace])

fig.update_layout(
    title='Water Transport CO₂ Emissions vs Sea Level Rise (1995–2018)',
    xaxis=dict(title='Year'),

    yaxis=dict(
        title=dict(text='CO₂ Emissions (PPM)', font=dict(color='green')),
        range=[490, 900],
        tickfont=dict(color='green')
    ),

    yaxis2=dict(
        title=dict(text='Average Sea Level (mm)', font=dict(color='blue')),
        overlaying='y',
        side='right',
        tickfont=dict(color='blue')
    ),

    legend=dict(x=0.5, y = -0.2, xanchor='center', orientation='h')
)


fig.show()
```

```python
year_cols = [str(y) for y in range(1995, 2019)]

# Total emissions across all industries
total = industry_df[year_cols].sum()

# Water transport emissions
water = industry_df[industry_df['Industry'] == 'Water transport'][year_cols].iloc[0]

# Compute proportion (%)
proportion_df = pd.DataFrame({
    'Year': [int(y) for y in year_cols],
    'Proportion': (water / total) * 100
})
proportion_df['RelativeGrowth'] = (proportion_df['Proportion'] / proportion_df['Proportion'].iloc[0]) * 100
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=proportion_df['Year'],
    y=proportion_df['RelativeGrowth'],
    mode='lines+markers',
    name='Relative Growth',
    line=dict(color='green'),
    marker=dict(size=6),
    hovertemplate='Year: %{x}<br>Relative Growth: %{y:.2f}%',
))

fig.update_layout(
    title='Water Transport Emissions Share Growth (1995 Baseline = 100%)',
    xaxis_title='Year',
    yaxis_title='Relative Increase Since 1995 (%)',
    legend=dict(x=0.5, xanchor='center', orientation='h'),
    hovermode='x unified'
)

fig.show()

```

```python
# Plot emission (ppm) per million dollars due to water transport per year
emission_per_million_df = water_transport_df.iloc[:1]

emission_per_million_df = emission_per_million_df.melt(
    var_name='Year',
    value_name='emission_per_million'
)
emission_per_million_df['Year'] = emission_per_million_df['Year'].astype(int)
alt.Chart(emission_per_million_df).mark_line(point=True).encode(
    x=alt.X('Year:O', title='Year', axis=alt.Axis(labelAngle=0)),
    y=alt.Y('emission_per_million:Q', title='CO₂ Emissions per $1M Output'),
    tooltip=['Year', 'emission_per_million']
).properties(
    title='Water Transport CO₂ emission_per_million (1995–2018)',
    width=600,
    height=400
).interactive()

alt.Chart(yearly_sea_levels).mark_line(point=True).encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('AvgSeaLevel:Q', title='Average Sea Level (mm)'),
    tooltip=['Year', 'AvgSeaLevel']
).properties(
    title='Average Global Sea Level Change by Year',
    width=600,
    height=400
).interactive()

```

```python
import altair as alt

# Line + area for emissions per $1M output (green)
emission_chart = alt.Chart(emission_per_million_df).mark_area(
    line={'color': 'green'},
    color='lightgreen',
    interpolate='monotone'
).encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('emission_per_million:Q', title='CO₂ Emissions per $1M Output'),
    tooltip=['Year', 'emission_per_million']
).properties(
    title='Water Transport CO₂ Emissions per $1M Output (1995–2018)',
)

# Line + area for sea levels (blue)
sea_level_chart = alt.Chart(yearly_sea_levels).mark_area(
    line={'color': 'blue'},
    color='lightblue',
    interpolate='monotone'
).encode(
    x=alt.X('Year:O', title='Year'),
    y=alt.Y('AvgSeaLevel:Q', title='Average Sea Level (mm)'),
    tooltip=['Year', 'AvgSeaLevel']
).properties(
    title='Average Global Sea Level Change by Year',
)

# Combine side-by-side
combined_chart = alt.hconcat(emission_chart, sea_level_chart)
combined_chart
```

```python
latest_emission = industry_df[industry_df['Unit'] == 'Millions of Metric tons of CO2'][['Industry', '1995', '2018']]
```

```python
latest_emission = latest_emission.groupby('Industry').sum().reset_index()
```

```python
latest_emission.sort_values('2018')
```

```python
import pandas as pd
import plotly.graph_objects as go

# Clean and convert emissions
latest_emission['2018'] = latest_emission['2018'].astype(float)

# Get top 5 emitters
top_emitters = latest_emission.sort_values('2018', ascending=False).head(5).copy()

# Add Water Transport row manually if not in top 5
water_row = latest_emission[latest_emission['Industry'].str.lower().str.contains('water transport')]

if not water_row.empty and not water_row['Industry'].iloc[0] in top_emitters['Industry'].values:
    top_emitters = pd.concat([top_emitters, water_row])

# Sort for clean plotting
top_emitters = top_emitters.sort_values('2018', ascending=True)

# Highlight water transport
colors = ['steelblue' if 'water transport' in ind.lower() else 'lightgray' for ind in top_emitters['Industry']]

# Create bar chart
fig = go.Figure(go.Bar(
    x=top_emitters['2018'],
    y=top_emitters['Industry'],
    orientation='h',
    marker_color=colors,
    hovertemplate='%{y}<br>Emissions: %{x:.2f} MtCO₂'
))

fig.update_layout(
    title='Top CO₂ Emitting Industries in 2018 (with Water Transport)',
    xaxis_title='CO₂ Emissions (Million Metric Tons)',
    yaxis_title='Industry',
    width=700,
    height=400
)

fig.show()
```

```python
# Ensure emissions are float
latest_emission['1995'] = latest_emission['1995'].astype(float)
latest_emission['2018'] = latest_emission['2018'].astype(float)

# Get top industries for both years (or just all of them)
df_1995 = latest_emission[['Industry', '1995']].copy()
df_1995.columns = ['Industry', 'Emissions']
df_1995['Year'] = 1995

df_2018 = latest_emission[['Industry', '2018']].copy()
df_2018.columns = ['Industry', 'Emissions']
df_2018['Year'] = 2018
```

```python
# Tag water transport to color it differently
def tag_category(industry):
    return 'Water Transport' if 'water transport' in industry.lower() else 'Other'

df_1995['Highlight'] = df_1995['Industry'].apply(tag_category)
df_2018['Highlight'] = df_2018['Industry'].apply(tag_category)
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Make pie charts clean: show hover only, no visual text
fig = make_subplots(rows=1, cols=2, specs=[[{'type': 'domain'}, {'type': 'domain'}]],
                    subplot_titles=['1995', '2018'])

# Pie for 1995
fig.add_trace(go.Pie(
    labels=df_1995['Industry'],
    values=df_1995['Emissions'],
    marker_colors=[
        'steelblue' if h == 'Water Transport' else 'lightgray'
        for h in df_1995['Highlight']
    ],
    textinfo='none',  # 🔥 hide all labels
    hovertemplate='%{label}: %{percent}',
    name='1995'
), row=1, col=1)

# Pie for 2018
fig.add_trace(go.Pie(
    labels=df_2018['Industry'],
    values=df_2018['Emissions'],
    marker_colors=[
        'steelblue' if h == 'Water Transport' else 'lightgray'
        for h in df_2018['Highlight']
    ],
    textinfo='none',
    hovertemplate='%{label}: %{percent}',
    name='2018'
), row=1, col=2)

# Layout
fig.update_layout(
    title_text='CO₂ Emissions by Industry (1995 vs 2018)',
    width=900,
    height=500,
    showlegend=False
)

# Add floating annotation for Water Transport in 1995
fig.add_annotation(
    x=0.07, y=0.4,
    text=f"Water Transport: 3.13%",
    showarrow=True,
    font=dict(size=13, color='steelblue'),
    bgcolor='white',
    opacity=0.9
)
# Add floating annotation for Water Transport in 1995
fig.add_annotation(
    x=0.6, y=0.5,
    text=f"Water Transport: 3.2%",
    showarrow=True,
    font=dict(size=13, color='steelblue'),
    bgcolor='white',
    opacity=0.9
)
fig.show()
```
